```python
def get_movies(params=None):
    """
    page - страница
    limit - количество
    year - год/диапазон
    raiting.kp - рейтинг
    genres.name - жанр
    """
    url = f"{BASE_URL}/v1.4/movie"
    response = requests.get(url, headers=headers, params=params)
    if (response.status_code == 200):
        return response.json
    else:
        print(f"Ошибка при получении данных {response.status_code}")
        return None
```

# Работа с апи

```python
params = {
    "page": 1,
    "limit": 250,
    "notNullFields": [
        "description",
        "name",
        "year", 
        "rating.kp",
        "genres.name"
    ]
}
```

```python
data = get_movies(params)
```

```python
df = pd.DataFrame(data()['docs'])
```

```python
df.head()
```

```python
df_noorm = df[['id', 'name', 'alternativeName', 'year', 'description', 'movieLength']].copy()
```

```python
df_noorm['rairing'] = df['votes'].apply(lambda x: x.get('kp') if pd.notna(x) else None)
```

```python
df_noorm['rating'] = df['rating'].apply(lambda x: x.get('kp') if pd.notna(x) else None)
```

```python
df_noorm['genres'] = df['genres'].apply(
    lambda x: ', '.join([g['name'] for g in x]) if isinstance(x, list) else ''
)
```

```python
df_noorm.head(250)
```

```python
df_with_description = df_noorm[df_noorm['description'].notna() & (df_noorm['description'] != '')].copy()
```
